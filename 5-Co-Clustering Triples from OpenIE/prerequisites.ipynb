# Conceptual Prerequisites for `Co-Clustering Triples from Open Information Extraction`

## 1. Few Important Clustering Algorithms

### 1.1 K-Means Clustering Algorithm
K-Means is an unsupervised learning algorithm used to group data into k distinct clusters, where each data point belongs to the cluster with the nearest mean (called the centroid).

#### **Intuition**
Imagine we are sorting a box of colored beads we are colorblind. We can’t label them, but we can see some similarity between them — maybe based on brightness and size.
We want to:
- Group them into k piles such that beads in the same pile are similar.
- Choose representative beads (centroids) that describe each pile.
- Iterate until you’re confident your grouping makes sense.
